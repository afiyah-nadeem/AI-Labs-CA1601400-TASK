```python
# =======================================================
# Lab 4: A* Search, Admissibility, and Heuristic Audit
# =======================================================
import heapq

# Section 4.2 Graph
graph = {
    "S": [("A", 1), ("B", 4)],
    "A": [("C", 2)],
    "B": [("G", 5)],
    "C": [("G", 3)],
    "G": []
}

# Heuristics
h_admissible = {"S": 5, "A": 4, "B": 4, "C": 2, "G": 0}
h_ucs = {k: 0 for k in h_admissible}
h_overestimate = {"S": 5, "A": 4, "B": 4, "C": 10, "G": 0}  # h(C) modified to 10

def search_astar(h_func):
    pq = [(h_func["S"], 0, "S", ["S"])]
    best = {"S": 0}
    expanded_nodes = 0
    trace = []

    while pq:
        f, g, node, path = heapq.heappop(pq)

        # Log popped node for trace
        trace.append((node, g, h_func[node], f))

        if node == "G":
            return g, path, expanded_nodes, trace

        if g > best.get(node, float("inf")):
            continue

        expanded_nodes += 1  # Node expanded only when examining successors

        for nxt, w in graph.get(node, []):
            ng = g + w
            if ng < best.get(nxt, float("inf")):
                best[nxt] = ng
                heapq.heappush(pq, (ng + h_func[nxt], ng, nxt, path + [nxt]))

    return float("inf"), None, expanded_nodes, trace

# Execute experiments
cost_astar, path_astar, exp_astar, trace_astar = search_astar(h_admissible)
cost_ucs, path_ucs, exp_ucs, trace_ucs = search_astar(h_ucs)
cost_over, path_over, exp_over, trace_over = search_astar(h_overestimate)

print("--- Lab 4 Task 2 & 3: Comparison Summary ---")
print(f"{'Search Configuration':<32} | {'Cost':<5} | {'Expanded':<8} | {'Path Found'}")
print("-" * 75)
print(f"{'A* (Admissible Heuristic)':<32} | {cost_astar:<5} | {exp_astar:<8} | {' -> '.join(path_astar)}")
print(f"{'UCS Control (h = 0)':<32} | {cost_ucs:<5} | {exp_ucs:<8} | {' -> '.join(path_ucs)}")
print(f"{'A* (Overestimated h(C)=10)':<32} | {cost_over:<5} | {exp_over:<8} | {' -> '.join(path_over)}")

print("\n--- Lab 4 Task 1: First 3 Expansions Trace (A* Admissible) ---")
print(f"{'Step':<5} | {'Node':<5} | {'g(n)':<5} | {'h(n)':<5} | {'f(n) = g + h'}")
print("-" * 45)
for i, step in enumerate(trace_astar[:3], 1):
    print(f"{i:<5} | {step[0]:<5} | {step[1]:<5} | {step[2]:<5} | {step[3]}")
```
