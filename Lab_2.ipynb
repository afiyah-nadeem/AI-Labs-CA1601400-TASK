```python
# =======================================================
# Lab 2: Stateful Smart Classroom Agent
# =======================================================

class StatefulClassroomAgent:
    def __init__(self):
        self.previous_mode = None  # Internal state storage

    def act(self, temp, occupied):
        # Priority 1: Vacant room goes to ECO
        if not occupied:
            target_mode = "ECO"
        # Priority 2: Temperature thresholds for occupied room
        elif temp > 26:
            target_mode = "COOL"
        elif temp < 20:
            target_mode = "WARM"
        else:
            target_mode = "IDLE"

        # Determine whether to transmit command (avoid redundant transmissions)
        command_sent = (target_mode != self.previous_mode)
        last_mode = self.previous_mode
        self.previous_mode = target_mode

        return last_mode, target_mode, command_sent

agent = StatefulClassroomAgent()

# Required test cases: repeated percepts, exact boundaries, mode transitions, empty room
percept_stream = [
    (29, True),   # Initial COOL
    (29, True),   # Repeated percept (command suppressed)
    (26, True),   # Upper boundary (transitions to IDLE)
    (20, True),   # Lower boundary (remains IDLE)
    (19, True),   # Mode transition to WARM
    (29, False)   # Empty room priority over high temp (ECO)
]

print("--- Lab 2 Task 3: Stateful Agent Execution Trace ---")
print(f"{'Step':<5} | {'Percept (T, Occ)':<18} | {'Prev Mode':<10} | {'Target Mode':<12} | {'Transmit?'}")
print("-" * 65)

for step, percept in enumerate(percept_stream, 1):
    prev, target, sent = agent.act(*percept)
    print(f"{step:<5} | {str(percept):<18} | {str(prev):<10} | {target:<12} | {sent}")
```
