```python
# =======================================================
# Lab 1: Symbolic vs. Probabilistic Decision Systems
# =======================================================

# Task 1 & 1.5: Starter Comparison
def symbolic(score, complete):
    return "Review" if complete and score >= 70 else "Hold"

def threshold(probability):
    return "Review" if probability >= 0.70 else "Hold"

print("--- Lab 1.5 Starter Disagreement Table ---")
cases = [
    ("complete_high", 82, True, 0.81),
    ("complete_low", 68, True, 0.74),
    ("incomplete_high", 91, False, 0.88),
    ("boundary", 70, True, 0.70)
]

print(f"{'Case Name':<18} | {'Symbolic':<10} | {'Threshold':<10} | {'Disagreement'}")
print("-" * 55)
for name, score, complete, prob in cases:
    a, b = symbolic(score, complete), threshold(prob)
    print(f"{name:<18} | {a:<10} | {b:<10} | {a != b}")

# Task 2: Extended Transparent Rule System
# Added Conditions: minimum_attendance >= 75 and disciplinary_flag == False
def extended_rule_system(score, complete, attendance, disciplinary_flag):
    if disciplinary_flag:
        return "Hold", "Disciplinary flag active"
    if not complete:
        return "Hold", "Incomplete documentation"
    if score < 70:
        return "Hold", "Academic score below 70"
    if attendance < 75:
        return "Hold", "Attendance below 75%"
    return "Review", "All criteria satisfied"

print("\n--- Lab 1 Task 2: Extended Rules (4 Core Test Cases) ---")
test_cases_extended = [
    ("All Criteria Met", 85, True, 88, False, "Review"),
    ("Low Attendance", 85, True, 60, False, "Hold"),
    ("Disciplinary Flag", 92, True, 90, True, "Hold"),
    ("Boundary Score (70)", 70, True, 75, False, "Review")
]

print(f"{'Test Case':<22} | {'Expected':<8} | {'Actual':<8} | {'Reason'}")
print("-" * 65)
for name, sc, comp, att, disc, expected in test_cases_extended:
    actual, reason = extended_rule_system(sc, comp, att, disc)
    print(f"{name:<22} | {expected:<8} | {actual:<8} | {reason}")
```
